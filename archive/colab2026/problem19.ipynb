## Least Squares Regression

Let $n \geq d$.
Consider a full-rank, feature matrix $\mathbf{A} \in \mathbb{R}^{n \times d}$ and a target vector $\mathbf{b} \in \mathbb{R}^n$.
The regression problem is to find a minimizing vector
\begin{align*}
\mathbf{x}^* = \arg \min_{\mathbf{x} \in \mathbb{R}^d}
\| \mathbf{Ax} - \mathbf{b} \|_2^2.
\end{align*}

In class, we showed how to quickly solve the regression problem by projecting our feature matrix and target vector into a lower dimensional space.
In this problem, you will derive the exact solution for $\mathbf{x}^*$ and explore how long it takes to compute it.

### Part A: Optimal Solution

Show that $\mathbf{x}^* = \left(\mathbf{A}^\top \mathbf{A} \right)^{-1} \mathbf{A}^\top \mathbf{b}$.

**Hint:** Compute the partial derivative with respect to $x_k$ of the $\ell_2$-norm and set it equal to 0.

### Part B: Time Complexity

Explain why computing $\left(\mathbf{A}^\top \mathbf{A} \right)^{-1} \mathbf{A}^\top \mathbf{b}$ takes $O(nd^2)$ time. Then, load a regression dataset of your choice (e.g., from ```scikit-learn```), build $\mathbf{A}$ and $\mathbf{b}$ and find the optimal solution $\mathbf{x}^*$.
