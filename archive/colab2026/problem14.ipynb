## Visual Applications of SVD

### Part A: Low Rank Approximations to Images

Load an image of your choice from the internet as a matrix $\mathbf{X} \in \mathbb{R}^{n \times d}$. Use ```numpy```'s built-in tools to compute its SVD. Plot the singular values by their index; this is known as a *spectrum* plot.

Next, find plot the best rank-$k$ approximation for $k=1, 5, 10, 50, 100, 500, 1000$. How do the rank-$k$ approximations relate to the spectrum?
