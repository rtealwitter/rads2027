##  Set Size Estimate Accuracy

Recall the set size estimation problem, where we draw $m$ uniform samples from a set of size $n$.
Let $D$ be the number of duplicates in these samples that we observed.
On the notes, we claimed that:

If we make $m \geq c \frac{\sqrt{n}}{\epsilon}$ samples for a particular constant $c$, then the estimate $\widehat{n} = \frac{m(m-1)}{2D}$ satisfies $(1-\epsilon) n \leq \widehat{n} \leq (1+\epsilon) n$ with probability $9/10$.
