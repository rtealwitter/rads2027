##  Wikipedia Articles

In class, we calculated the number of duplicates in a sample of size $m$ as
$$
D = \sum_{i=1}^m \sum_{j=i+1}^m D_{i,j}
$$
where $D_{i,j}$ is an indicator random variable that the $i$th and $j$th sampled items are the same.
In expectation when the samples are drawn uniformly at random, we saw that
$$
\mathbb{E}[D] = \frac{m (m-1)}{2 n}.
$$

### Part A

In practice, we know $m$ the number of samples we've taken and $D$ the number of duplicates in the sample.
Using these quantities, suggest a method for estimating $n$---the set size---inspired by our expression for $\mathbb{E}[D]$.

### Part B

Implement your method from Part A to estimate the number of unique articles in the English Wikipedia.
You can access ``random'' articles (see the discussion [here](https://en.wikipedia.org/wiki/Wikipedia:FAQ/Technical#random)) by visiting the link: https://en.wikipedia.org/wiki/Special:Random.

According to the article [here](https://en.wikipedia.org/wiki/English_Wikipedia), English Wikipedia has 6.7 million articles. How does that compare to your estimate? How does the fact that the random article feature doesn't perfectly return random articles bias your estimate?

Here's some code you can use:

```python
import requests
import math

# Increase samples to get enough collisions
m = 100000
batch_size = 500
titles = []

URL = "https://en.wikipedia.org/w/api.php"
headers = {"User-Agent": "WikiSizeEstimator/1.0"}

print(f"Fetching {m} random URLs (this will take a few minutes)...")

while len(titles) < m:
    params = {
        "action": "query",
        "format": "json",
        "list": "random",
        "rnlimit": batch_size,
        "rnnamespace": 0
    }

    try:
        response = requests.get(url=URL, params=params, headers=headers)
        data = response.json()

        if "query" in data:
            for page in data["query"]["random"]:
                titles.append(page['id']) # Use ID, it's safer/faster than title strings
        else:
            print("Retrying...")

    except Exception:
        pass
```
